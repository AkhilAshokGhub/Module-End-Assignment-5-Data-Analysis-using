```python
columns = [
    'likes',
    'comments',
    'shares',
    'watch_time_sec',
    'engagement_rate',
    'follower_count'
]

for col in columns:

    print("\nColumn:", col)

    print("Mean:", sme[col].mean())

    print("Median:", sme[col].median())

    print("Mode:", sme[col].mode().tolist())
```

```python
for col in columns:

    print("\nColumn:", col)

    print("Standard Deviation:", sme[col].std())

    print("Variance:", sme[col].var())
```

```python
for col in columns:

    print("\nColumn:", col)

    print("25th Percentile:",
          sme[col].quantile(0.25))

    print("50th Percentile:",
          sme[col].quantile(0.50))

    print("75th Percentile:",
          sme[col].quantile(0.75))

    print("90th Percentile:",
          sme[col].quantile(0.90))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.scatter(sme['likes'], sme['impression_count'], alpha=0.5)
plt.title('Like vs Impression')
plt.xlabel('Impression')
plt.ylabel('Like')
plt.show()
```

```python
plt.plot(sme['posted_at'], sme['engagement_rate'], alpha=0.5)
plt.title('Like vs Impression')
plt.xlabel('engagement_rate')
plt.ylabel('posted_at')
plt.show()
```

```python
daily_engagement = sme.groupby('posted_at')[
    'engagement_rate'
].mean()
plt.figure(figsize=(20, 5))
plt.plot(
    daily_engagement.index,
    daily_engagement.values
)
plt.xlabel('Date')
plt.ylabel('Average Engagement Rate')
plt.title('Daily Engagement Trend')
plt.show()

```

```python
category_count = sme['post_category'].value_counts()

plt.figure(figsize=(9, 5))

plt.bar(
    category_count.index,
    category_count.values
)

plt.xlabel('Category')

plt.ylabel('Number of Posts')

plt.title('Posts by Category')

plt.xticks(rotation=45)

plt.show()
```

```python
gender_count = sme['gender'].value_counts()

plt.figure(figsize=(12, 7))

plt.pie(
    gender_count.values,
    labels=gender_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Gender Distribution')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    sme['age'].dropna(),
    bins=15,
    edgecolor='black'
)

plt.xlabel('Age')

plt.ylabel('Frequency')

plt.title('Age Distribution')

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=sme,
    x='post_category',
    y='likes',
    errorbar=None
)

plt.title('Average Likes by Category')

plt.xlabel('Category')

plt.ylabel('Average Likes')

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=sme,
    x='sentiment',
    y='follower_count'
)

plt.title('Follower Count vs Sentiment')

plt.xlabel('Sentiment')

plt.ylabel('Followers')

plt.show()
```

```python
post_type_performance = sme.groupby('post_type').agg(
    avg_engagement=('engagement_rate', 'mean'),
    avg_likes=('likes', 'mean'),
    avg_comments=('comments', 'mean'),
    avg_shares=('shares', 'mean')
).sort_values(by='avg_engagement', ascending=False)

print(post_type_performance)
```

```python
category_performance = sme.groupby('post_category').agg(
    avg_engagement=('engagement_rate', 'mean'),
    avg_likes=('likes', 'mean'),
    avg_shares=('shares', 'mean')
).sort_values(by='avg_engagement', ascending=False)

print(category_performance)
```

```python
country_performance = sme.groupby('country').agg(
    avg_engagement=('engagement_rate', 'mean'),
    total_posts=('engagement_rate', 'count')
).sort_values(by='avg_engagement', ascending=False)

print(country_performance.head(10))
```

```python
sme['age_group'] = pd.cut(
    sme['age'],
    bins=[0, 18, 25, 35, 45, 55, 100],
    labels=['Under 18', '18-25', '26-35', '36-45', '46-55', 'Above 55']
)

age_performance = sme.groupby('age_group', observed=False).agg(
    avg_engagement=('engagement_rate', 'mean'),
    avg_likes=('likes', 'mean'),
    total_posts=('engagement_rate', 'count')
)

print(age_performance)
```

```python
verified_performance = sme.groupby('is_verified').agg(
    avg_engagement=('engagement_rate', 'mean'),
    avg_likes=('likes', 'mean'),
    avg_comments=('comments', 'mean'),
    total_posts=('engagement_rate', 'count')
)

print(verified_performance)
```

```python
sme['posted_at'] = pd.to_datetime(sme['posted_at'], errors='coerce')

sme['post_hour'] = sme['posted_at'].dt.hour

hour_performance = sme.groupby('post_hour').agg(
    avg_impressions=('impressions', 'mean'),
    total_posts=('impressions', 'count')
).sort_values(by='avg_impressions', ascending=False)

print(hour_performance)
```

```python
device_performance = sme.groupby('device_type').agg(
    avg_watch_time=('watch_time_sec', 'mean'),
    avg_engagement=('engagement_rate', 'mean'),
    total_posts=('watch_time_sec', 'count')
).sort_values(by='avg_watch_time', ascending=False)

print(device_performance)
```

```python
sentiment_performance = sme.groupby('sentiment').agg(
    avg_engagement=('engagement_rate', 'mean'),
    avg_likes=('likes', 'mean'),
    avg_comments=('comments', 'mean'),
    avg_shares=('shares', 'mean'),
    total_posts=('engagement_rate', 'count')
).sort_values(by='avg_engagement', ascending=False)

print(sentiment_performance)
```

```python
negative_neutral = sme[sme['sentiment'].isin(['negative', 'neutral'])]

sentiment_comparison = negative_neutral.groupby('sentiment').agg(
    avg_engagement=('engagement_rate', 'mean'),
    avg_likes=('likes', 'mean'),
    avg_comments=('comments', 'mean'),
    avg_shares=('shares', 'mean'),
    avg_watch_time=('watch_time_sec', 'mean'),
    total_posts=('sentiment', 'count')
)

print(sentiment_comparison)
```
