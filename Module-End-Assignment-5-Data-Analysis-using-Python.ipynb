```python
sme['likes'] = pd.to_numeric(sme['likes'], errors= 'coerce')
```

```python
sme['posted_at'] = pd.to_datetime(sme['posted_at'],format= '%d-%m-%y', errors='coerce')
```

```python
likes_array = np.array(sme['likes'])
print(likes_array)
```

```python
new_likes = likes_array + 10
print(new_likes)

double_likes = likes_array * 2
print(double_likes)

avg_likes = np.nanmean(likes_array)
print(avg_likes)
```

```python
max_likes = np.nanmax(likes_array)
min_likes = np.nanmin(likes_array)
print(max_likes)
print(min_likes)

print(np.nansum(likes_array))
```

```python
# ❖	Task 2 — Data Cleaning
sme.isnull().sum()
```

```python
sme.isna().sum().sum()
```

```python
sme_drop = sme.dropna
print(sme_drop)
```

```python
sme.isnull().sum()
```

```python
sme['age'] = sme["age"].fillna(38)
sme['likes'] = sme['likes'].fillna(10107.0)
sme['comments'] = sme['comments'].fillna(1502.1)
sme['shares'] = sme['shares'].fillna(1002)
```

```python
sme['gender'] = sme['gender'].fillna(sme['gender'].mode()[0])
sme['sentiment'] = sme['sentiment'].fillna(sme['sentiment'].mode()[0])
```

```python
sample = pd.Series(['positive', np.nan, 'negative', np.nan])

print(sample.ffill())

sme['sentiment'] = sme['sentiment'].ffill()
```

```python
sample = pd.Series(['positive', np.nan, 'negative', np.nan])

print(sample.bfill())
```
